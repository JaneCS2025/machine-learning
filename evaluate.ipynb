```python
import numpy
import matplotlib.pyplot as plt
numpy.random.seed(2)
```

```python
# number of minutes before making a purchase
x = numpy.random.normal(3, 1, 100) # mean value. std, 100 people
```

```python
# amount of money spent on purchase
y = numpy.random.normal(150, 40, 100) / x
```

```python
plt.scatter(x, y)
plt.show()
```

```python
# Train/Test 
train_x = x[:80]
train_y = y[:80]

test_x = x[80:]
test_y = y[80:]
```

```python
# Display the Training Set
plt.scatter(train_x, train_y)
```

```python
plt.scatter(test_x, test_y)
```

```python
# create the model using training set

mymodel = numpy.poly1d(numpy.polyfit(train_x, train_y, 4))
myline = numpy.linspace(0, 6, 100)

plt.scatter(train_x, train_y)
plt.plot(myline, mymodel(myline))
```

```python
# R2 - R-squared
# r2_score() that will help us find this relationship
# value is between 0 - 1, 0 means no relationship, 1 means totally related

from sklearn.metrics import r2_score

r2 = r2_score(train_y, mymodel(train_x))
r2_test = r2_score(test_y, mymodel(test_x))
print(r2)
print(r2_test)
```
