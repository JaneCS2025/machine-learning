```python
import pandas
df = pandas.read_csv("data/data_test.csv")
df.head()
```

```python
# data pre-processing 
d = {
  'UK': 0,
  'USA':1,
   'N': 2 
}

df['Nationality'] = df['Nationality'].map(d)

d = {
    'YES': 1,
    'NO': 0
}
df['Go'] = df['Go'].map(d)
df.head()
```

```python
filterd_df = df[(df['Rank'] <=8.5) & (df['Rank'] >=6.5)]
print(filterd_df)
```

```python
features = ['Age', 'Experience', 'Rank', 'Nationality']

X = df[features]
y = df['Go']
X.head()
print(df)
```

```python
from sklearn import tree
from sklearn.tree import DecisionTreeClassifier
import matplotlib.pyplot as plt
```

```python
dtree = DecisionTreeClassifier()
dtree = dtree.fit(X, y)
tree.plot_tree(dtree, feature_names=features)
```
