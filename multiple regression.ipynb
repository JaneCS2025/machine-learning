```python
import pandas
from sklearn import linear_model

df = pandas.read_csv("data/data.csv")

X = df[['Weight', 'Volume']]
y = df['CO2']

regr = linear_model.LinearRegression()
regr.fit(X, y)

print(regr.coef_)
```

```python
predictedCO2 = regr.predict([[3300, 1300]]) # 3300 kg, 1300 Cubic meters
print(predictedCO2)
```
